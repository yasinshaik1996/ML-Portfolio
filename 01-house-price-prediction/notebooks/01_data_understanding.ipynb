```python
#ColumnTransformer → controls WHICH columns get processed HOW

#Pipeline          → controls WHICH steps happen in WHAT ORDER

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
categorical_pipeline = Pipeline(
    steps=[
        ("imputer", categorical_imputer),
        ("encoder", categorical_encoder)
    ]
)
#to check the pipeline
categorical_pipeline
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", numeric_imputer),
        ("scaler", scaler)
    ]
)

numeric_pipeline
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

preprocessor
```

```python
preprocessor.fit(X_train)
```

```python
# Linear Regression
#       ↓
# Minimize prediction error
##############################################
# Ridge Regression
#        ↓
# Minimize prediction error
#        +
# Penalize large coefficients

#                  ┌── Numerical → Imputer → Scaler ──┐
# Raw X ───────────┤                                    ├──→ Ridge
#                  └── Categorical → Imputer → Encoder ┘


#                  Validation
#                  R²
#                    │
#        ┌───────────┴───────────┐
#        │                       │
# Linear Regression         Ridge Regression
#      0.8687                     ?

#Scaling means: Transforming numerical features so that they are on a comparable scale.

# fit() Means: Learn the required information from the data.

# For StandardScaler, that means learning things like:
# mean
# standard deviation
```

```python
# alpha = 0
#      ↓
# almost ordinary Linear Regression

# small alpha
#      ↓
# small penalty

# large alpha
#      ↓
# stronger penalty
#      ↓
# coefficients pushed closer to zero

# Finding the best value of a model's hyperparameter is called: Hyperparameter Tuning
# And alpha is Ridge's main hyperparameter

from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=10)
ridge_model.fit(X_train_processed, y_train)
y_pred_ridge = ridge_model.predict(X_valid_processed)
print("Prediction shape:", y_pred_ridge.shape)
```

```python
from sklearn.metrics import mean_absolute_error

mae_ridge = mean_absolute_error(y_valid, y_pred_ridge)

print("Ridge MAE:", mae_ridge)
```

```python
from sklearn.metrics import mean_squared_error

mse_ridge = mean_squared_error(y_valid, y_pred_ridge)

print("Ridge MSE:", mse_ridge)
```

```python
import numpy as np

rmse_ridge = np.sqrt(mse_ridge)

print("Ridge RMSE:", rmse_ridge)
```

```python
from sklearn.metrics import r2_score

r2_ridge = r2_score(y_valid, y_pred_ridge)

print("Ridge R²:", r2_ridge)
```

```python
#cross-validation
from sklearn.model_selection import KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
for fold, (train_index, val_index) in enumerate(kf.split(X_train), start=1):
    print(
        f"Fold {fold}: "
        f"Training = {len(train_index)}, "
        f"Validation = {len(val_index)}"
    )
```

```python
from sklearn.pipeline import Pipeline

ridge_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", Ridge())
    ]
)

ridge_pipeline
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    "model__alpha": [0.1, 1, 10, 50, 100, 500, 1000]
}
grid_search = GridSearchCV(
    estimator=ridge_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)
grid_search.fit(X_train, y_train)
```

```python
print("Best alpha:", grid_search.best_params_)
```

```python
cv_results = pd.DataFrame(grid_search.cv_results_)

cv_results[
    [
        "param_model__alpha",
        "mean_test_score",
        "std_test_score"
    ]
]
```

```python
cv_results["mean_MAE"] = -cv_results["mean_test_score"]

cv_results[
    [
        "param_model__alpha",
        "mean_MAE",
        "std_test_score"
    ]
].sort_values("mean_MAE")
```

```python
best_ridge = grid_search.best_estimator_

print(best_ridge)
```

```python
y_valid_pred_best = best_ridge.predict(X_valid)
```

```python
from sklearn.metrics import mean_absolute_error

final_mae = mean_absolute_error(
    y_valid,
    y_valid_pred_best
)

print("Final Ridge MAE:", final_mae)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

final_mse = mean_squared_error(
    y_valid,
    y_valid_pred_best
)

final_rmse = np.sqrt(final_mse)

final_r2 = r2_score(
    y_valid,
    y_valid_pred_best
)

print("Final Ridge MSE:", final_mse)
print("Final Ridge RMSE:", final_rmse)
print("Final Ridge R²:", final_r2)
```
